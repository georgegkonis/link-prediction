```python
import pandas as pd

# Load the dataset
data = pd.DataFrame("data/links.csv")

# Calculate total number of links
total_links = len(data)

# Calculate the number of unique nodes
unique_nodes = len(set(data['id1']).union(set(data['id2'])))

# Calculate average links per node
average_links = total_links / unique_nodes

# Calculate number of links for each label
label_counts = data['label'].value_counts()

# Display the results
total_links, unique_nodes, average_links, label_counts
```
